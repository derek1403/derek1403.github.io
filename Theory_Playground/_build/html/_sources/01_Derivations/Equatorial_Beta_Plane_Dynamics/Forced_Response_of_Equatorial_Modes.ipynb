# Forced Response of Equatorial Modes (赤道模態的受迫解：一行除法)

## 證明目標:

**端點①的核心。** 用 [赤道波本徵函數](Equatorial_Wave_Eigenfunctions.ipynb) 當基底做正規模態轉換，
把 [淺水系統](Fourier_Transform_to_Shallow_Water_System.ipynb) 那個常微分方程組**塌縮成一行純代數的除法**。

* (a) 正規模態轉換對 —— 展開式：

$$\hat{\boldsymbol{\eta}}_m(\hat{y}) = \sum_{n = -1}^{\infty}\sum_{r}\hat{\eta}_{mnr}\,\mathbf{K}_{mnr}(\hat{y})$$

* (b) 正規模態轉換對 —— 取係數（只要做一次內積）：

$$\hat{\eta}_{mnr} = \left(\hat{\boldsymbol{\eta}}_m(\hat{y}),\ \mathbf{K}_{mnr}(\hat{y})\right)$$

* (c) **★ 受迫解 —— 常微分方程組塌縮成一行除法：**

$$\hat{\eta}_{mnr} = \frac{\kappa\hat{Q}_{mnr}}{\alpha + i\left(\nu_{mnr} - \dfrac{cm}{a}\right)}$$

  其中 $\hat{Q}_{mnr} = \left(\hat{\mathbf{Q}}_m,\ \mathbf{K}_{mnr}\right)$。

* (d) 共振條件 —— 當某個波模的相速接近對流的移速時，響應被放大到只受阻尼限制：

$$\left.\left|\hat{\eta}_{mnr}\right|\right|_{\nu_{mnr} = cm/a} = \frac{\kappa\left|\hat{Q}_{mnr}\right|}{\alpha}$$

* 註：(a)–(c) 就是 [Schubert & Masarik (2006)](../../06_References/Tropical_wave_dynamics/Schubert2006-Potential_vorticity_aspects_of_the_MJO/Potential_vorticity_aspects_of_the_MJO.ipynb) 的 $(4.20)$、$(4.19)$、$(4.21)$。
* 註：(c) 的物理讀法是**受迫阻尼振子的響應函數**：分子是強迫投影到該模態上的分量，分母的實部是阻尼、虛部是「該模態的自然頻率」與「強迫的移動頻率」之差。
* 註：強迫項 $\hat{Q}_{mnr}$ 的顯式形式需要先給定加熱的具體形狀，留到 [移動熱源的模態投影](Projection_of_a_Moving_Heat_Source.ipynb)。

## 假設與已知 (Assumptions & Preliminaries)

* **【已知 1】 [向量形式的淺水系統 (Vector form of the shallow water system)](https://derek1403.github.io/Theory_Playground/_build/html/01_Derivations/Equatorial_Beta_Plane_Dynamics/Fourier_Transform_to_Shallow_Water_System.html#e-proof-vector-form)：** 已於本庫 [Fourier Transform to the Shallow Water System](Fourier_Transform_to_Shallow_Water_System.ipynb)【證明 (e)】完整證明，此處直接引用不再重證

  $$\left(\alpha - \frac{imc}{a}\right)\hat{\boldsymbol{\eta}}_m + \mathcal{L}\hat{\boldsymbol{\eta}}_m = \kappa\hat{\mathbf{Q}}_m$$

  * $\hat{\boldsymbol{\eta}}_m$ : 第 $m$ 個緯向波數的狀態向量 (State vector) $[\text{依分量而定}]$
  * $\hat{\mathbf{Q}}_m$ : 強迫向量 (Forcing vector) $[\text{依分量而定}]$
  * $\mathcal{L}$ : 水平結構的線性算符 (Linear operator) $[\text{s}^{-1}]$
  * $\alpha$ : 常數阻尼率 (Constant damping rate) $[\text{s}^{-1}]$，$\alpha = \left(4 \ \text{days}\right)^{-1}$
  * $c$ : 對流包絡的東移速度 (Eastward propagation speed) $[\text{m}\cdot\text{s}^{-1}]$，$c = 5 \ \text{m}\cdot\text{s}^{-1}$
  * $m$ : 緯向波數 (Zonal wavenumber) $[\text{無單位}]$，整數
  * $a$ : 地球半徑 (Earth's radius) $[\text{m}]$
  * $\kappa$ : Poisson 常數 (Poisson constant) $[\text{無單位}]$

* **【已知 2】 [算符的反厄米性與本徵值 (Skew-Hermiticity and the eigenvalues)](https://derek1403.github.io/Theory_Playground/_build/html/01_Derivations/Equatorial_Beta_Plane_Dynamics/Energy_Inner_Product_and_Skew_Hermitian_Operator.html#d-proof-skew-hermiticity-of-the-operator)：** 已於本庫 [Energy Inner Product and the Skew-Hermitian Operator](Energy_Inner_Product_and_Skew_Hermitian_Operator.ipynb) 完整證明，此處直接引用不再重證

  * (a) 反厄米性：

    $$\left(\mathcal{L}\mathbf{f},\ \mathbf{g}\right) = -\left(\mathbf{f},\ \mathcal{L}\mathbf{g}\right)$$

  * (b) 本徵值為純虛數（$\nu_{mnr}$ 為實數）：

    $$\mathcal{L}\mathbf{K}_{mnr} = i\nu_{mnr}\mathbf{K}_{mnr}, \qquad \nu_{mnr} \in \mathbb{R}$$

  * (c) 內積的共軛線性性（第二個引數帶共軛）：

    $$\left(\mathbf{f},\ \sigma\mathbf{g}\right) = \sigma^{*}\left(\mathbf{f},\ \mathbf{g}\right)$$

  * $\mathbf{K}_{mnr}$ : 本徵函數 (Eigenfunction) $[\text{依分量而定}]$
  * $\nu_{mnr}$ : 本徵頻率 (Eigenfrequency) $[\text{s}^{-1}]$
  * $\sigma$ : 任意複數 (Arbitrary complex number) $[\text{無單位}]$
  * $\mathcal{L}$ : 水平結構的線性算符 (Linear operator) $[\text{s}^{-1}]$
  * $\mathbf{f},\ \mathbf{g}$ : 複數三分量向量函數 (Complex three-component vector functions) $[\text{依分量而定}]$

* **【已知 3】 [本徵函數的正交歸一性與完備性 (Orthonormality and completeness of the eigenfunctions)](https://derek1403.github.io/Theory_Playground/_build/html/01_Derivations/Equatorial_Beta_Plane_Dynamics/Equatorial_Wave_Eigenfunctions.html#d-proof-orthonormality)：** 正交歸一已於本庫 [Equatorial Wave Eigenfunctions](Equatorial_Wave_Eigenfunctions.ipynb)【證明 (d)】完整證明；完備性為反厄米算符的標準結果 [*Wu and Moore*, 2004]，此處一併引用不再重證

  * (a) 正交歸一：

    $$\left(\mathbf{K}_{mnr},\ \mathbf{K}_{mn'r'}\right) = \begin{cases}1, & \left(n', r'\right) = \left(n, r\right) \\ 0, & \left(n', r'\right) \neq \left(n, r\right)\end{cases}$$

  * (b) 完備性 —— 任一在該內積下平方可積的向量函數皆可展開：

    $$\hat{\boldsymbol{\eta}}_m(\hat{y}) = \sum_{n = -1}^{\infty}\sum_{r}\hat{\eta}_{mnr}\mathbf{K}_{mnr}(\hat{y})$$

  * $\hat{\eta}_{mnr}$ : 正規模態展開係數 (Normal mode expansion coefficient) $[\text{m}\cdot\text{s}^{-1}]$
  * $\mathbf{K}_{mnr}$ : 本徵函數 (Eigenfunction) $[\text{依分量而定}]$
  * $\hat{\boldsymbol{\eta}}_m$ : 第 $m$ 個緯向波數的狀態向量 (State vector) $[\text{依分量而定}]$
  * $\hat{y}$ : 無因次經向座標 (Dimensionless meridional coordinate) $[\text{無單位}]$

* **【假設 1】 展開式可逐項取內積 (Term-by-term inner product)：** 級數在該內積下收斂，故內積與求和可交換

  $$\left(\sum_{n,r}\hat{\eta}_{mnr}\mathbf{K}_{mnr},\ \mathbf{K}_{mn'r'}\right) = \sum_{n,r}\hat{\eta}_{mnr}\left(\mathbf{K}_{mnr},\ \mathbf{K}_{mn'r'}\right)$$

  * $\hat{\eta}_{mnr}$ : 正規模態展開係數 (Normal mode expansion coefficient) $[\text{m}\cdot\text{s}^{-1}]$
  * $m$ : 緯向波數 (Zonal wavenumber) $[\text{無單位}]$，整數
  * $n$ : 經向模態指標 (Meridional mode index) $[\text{無單位}]$
  * $r$ : 波型指標 (Wave-type index) $[\text{無單位}]$
  * $\mathbf{K}_{mnr}$ : 本徵函數 (Eigenfunction) $[\text{依分量而定}]$

* **【假設 2】 阻尼不為零 (Nonzero damping)：** 保證【證明 (c)】的分母不為零，受迫解處處良好定義

  $$\alpha > 0$$

  * $\alpha$ : 常數阻尼率 (Constant damping rate) $[\text{s}^{-1}]$，$\alpha = \left(4 \ \text{days}\right)^{-1}$
  * 註：這是**必要**的。$\alpha = 0$ 時，若強迫的移動頻率恰好等於某個模態的自然頻率（$\nu_{mnr} = cm/a$），分母歸零，響應發散 —— 那正是無阻尼共振。真實大氣中的輻射冷卻與摩擦提供了 $\alpha$，把共振峰壓成有限高度。

* **【定義 1】 強迫的模態投影 (Modal projection of the forcing)：** 強迫向量在第 $(m, n, r)$ 個本徵函數上的分量

  $$\hat{Q}_{mnr} \overset{\text{def}}{=} \left(\hat{\mathbf{Q}}_m,\ \mathbf{K}_{mnr}\right)$$

  * $\hat{Q}_{mnr}$ : 強迫的模態投影 (Modal projection of the forcing) $[\text{K}\cdot\text{s}^{-1}]$
  * $\hat{\mathbf{Q}}_m$ : 強迫向量 (Forcing vector) $[\text{依分量而定}]$
  * $\mathbf{K}_{mnr}$ : 本徵函數 (Eigenfunction) $[\text{依分量而定}]$

* **【推導 1】 反厄米性把算符換成本徵值 (Skew-Hermiticity converts the operator into its eigenvalue)：** 這是整篇的樞紐 —— $\mathcal{L}$ 作用在**未知**的 $\hat{\boldsymbol{\eta}}_m$ 上，經過反厄米性轉移到**已知**的 $\mathbf{K}_{mnr}$ 上

  $$\begin{gather*}
  \left(\mathcal{L}\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mnr}\right) &\overset{\text{已知 2(a)}}{=}& -\left(\hat{\boldsymbol{\eta}}_m,\ \mathcal{L}\mathbf{K}_{mnr}\right) \\
  &\overset{\text{已知 2(b)}}{=}& -\left(\hat{\boldsymbol{\eta}}_m,\ i\nu_{mnr}\mathbf{K}_{mnr}\right) \\
  &\overset{\text{已知 2(c)}}{=}& -\left(i\nu_{mnr}\right)^{*}\left(\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mnr}\right) \\
  &\overset{\text{已知 2(b)}}{=}& i\nu_{mnr}\left(\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mnr}\right)
  \end{gather*}$$

  * $\mathcal{L}$ : 水平結構的線性算符 (Linear operator) $[\text{s}^{-1}]$
  * $\hat{\boldsymbol{\eta}}_m$ : 第 $m$ 個緯向波數的狀態向量 (State vector) $[\text{依分量而定}]$
  * $m$ : 緯向波數 (Zonal wavenumber) $[\text{無單位}]$，整數
  * $\mathbf{K}_{mnr}$ : 本徵函數 (Eigenfunction) $[\text{依分量而定}]$
  * $n$ : 經向模態指標 (Meridional mode index) $[\text{無單位}]$
  * $r$ : 波型指標 (Wave-type index) $[\text{無單位}]$
  * $\nu_{mnr}$ : 本徵頻率 (Eigenfrequency) $[\text{s}^{-1}]$
  * $\hat{y}$ : 無因次經向座標 (Dimensionless meridional coordinate) $[\text{無單位}]$
  * $\hat{\eta}_{mnr}$ : 正規模態展開係數 (Normal mode expansion coefficient) $[\text{m}\cdot\text{s}^{-1}]$
  * $\kappa$ : Poisson 常數 (Poisson constant) $[\text{無單位}]$
  * $\hat{Q}_{mnr}$ : 強迫的模態投影 (Modal projection of the forcing) $[\text{K}\cdot\text{s}^{-1}]$
  * $\hat{\mathbf{Q}}_m$ : 強迫向量 (Forcing vector) $[\text{依分量而定}]$
  * $\alpha$ : 常數阻尼率 (Constant damping rate) $[\text{s}^{-1}]$，$\alpha = \left(4 \ \text{days}\right)^{-1}$
  * $c$ : 對流包絡的東移速度 (Eastward propagation speed) $[\text{m}\cdot\text{s}^{-1}]$，$c = 5 \ \text{m}\cdot\text{s}^{-1}$
  * $a$ : 地球半徑 (Earth's radius) $[\text{m}]$
  * 註：最後一步用到 $\nu_{mnr}$ **為實數**，故 $\left(i\nu_{mnr}\right)^{*} = -i\nu_{mnr}$。**兩個負號相乘變正號** —— 若 $\mathcal{L}$ 是厄米而非反厄米，這裡會多一個負號，後面的分母就湊不出「阻尼 ＋ 頻率差」的漂亮形式。

## 證明:

### (a) proof 正規模態展開式 (Normal mode expansion)

直接引用完備性。

$$\hat{\boldsymbol{\eta}}_m(\hat{y}) \overset{\text{已知 3(b)}}{=} \sum_{n = -1}^{\infty}\sum_{r}\hat{\eta}_{mnr}\,\mathbf{K}_{mnr}(\hat{y})$$

### (b) proof 展開係數的取法 (Extraction of the expansion coefficients)

把【證明 (a)】與 $\mathbf{K}_{mn'r'}$ 取內積，正交歸一性讓級數只剩一項。

$$\begin{gather*}
\left(\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mn'r'}\right) &\overset{\text{證明 (a)}}{=}& \left(\sum_{n = -1}^{\infty}\sum_{r}\hat{\eta}_{mnr}\mathbf{K}_{mnr},\ \mathbf{K}_{mn'r'}\right) \\
&\overset{\text{假設 1}}{=}& \sum_{n = -1}^{\infty}\sum_{r}\hat{\eta}_{mnr}\left(\mathbf{K}_{mnr},\ \mathbf{K}_{mn'r'}\right) \\
&\overset{\text{已知 3(a)}}{=}& \hat{\eta}_{mn'r'}
\end{gather*}$$

### (c) proof 受迫解 (Forced response)

把【已知 1】整條與 $\mathbf{K}_{mnr}$ 取內積：第一項由【證明 (b)】變成 $\hat{\eta}_{mnr}$，第二項由【推導 1】變成 $i\nu_{mnr}\hat{\eta}_{mnr}$，右端由【定義 1】變成 $\kappa\hat{Q}_{mnr}$。

$$\begin{gather*}
\kappa\hat{Q}_{mnr} &\overset{\text{定義 1}}{=}& \kappa\left(\hat{\mathbf{Q}}_m,\ \mathbf{K}_{mnr}\right) \\
\kappa\hat{Q}_{mnr} &\overset{\text{已知 1}}{=}& \left(\left(\alpha - \frac{imc}{a}\right)\hat{\boldsymbol{\eta}}_m + \mathcal{L}\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mnr}\right) \\
\kappa\hat{Q}_{mnr} &=& \left(\alpha - \frac{imc}{a}\right)\left(\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mnr}\right) + \left(\mathcal{L}\hat{\boldsymbol{\eta}}_m,\ \mathbf{K}_{mnr}\right) \\
\kappa\hat{Q}_{mnr} &\overset{\text{證明 (b),推導 1}}{=}& \left(\alpha - \frac{imc}{a}\right)\hat{\eta}_{mnr} + i\nu_{mnr}\hat{\eta}_{mnr} \\
\kappa\hat{Q}_{mnr} &=& \left[\alpha + i\left(\nu_{mnr} - \frac{cm}{a}\right)\right]\hat{\eta}_{mnr} \\
\hat{\eta}_{mnr} &\overset{\text{假設 2}}{=}& \frac{\kappa\hat{Q}_{mnr}}{\alpha + i\left(\nu_{mnr} - \dfrac{cm}{a}\right)}
\end{gather*}$$

### (d) proof 共振條件下的響應 (Response at resonance)

把【證明 (c)】取絕對值，分母是複數的模。

$$\begin{gather*}
\left|\hat{\eta}_{mnr}\right| &\overset{\text{證明 (c)}}{=}& \frac{\kappa\left|\hat{Q}_{mnr}\right|}{\left[\alpha^{2} + \left(\nu_{mnr} - \dfrac{cm}{a}\right)^{2}\right]^{1/2}} \\
\left.\left|\hat{\eta}_{mnr}\right|\right|_{\nu_{mnr} = cm/a} &=& \frac{\kappa\left|\hat{Q}_{mnr}\right|}{\left[\alpha^{2} + 0\right]^{1/2}} \\
\left.\left|\hat{\eta}_{mnr}\right|\right|_{\nu_{mnr} = cm/a} &\overset{\text{假設 2}}{=}& \frac{\kappa\left|\hat{Q}_{mnr}\right|}{\alpha}
\end{gather*}$$

## 物理解釋

### 一行除法背後的三層轉換

到這一步為止，原本的四維偏微分方程組已經被三層正交轉換壓成純代數：

| 層 | 工具 | 吃掉的維度 | 出處 |
|---|---|---|---|
| 第一層 | 垂直結構函數 $Z(z)$ | $z$ | [垂直分離](Separation_into_Horizontal_Structure_System.ipynb) |
| 第二層 | 緯向傅立葉 $e^{im\xi/a}$ | $\xi$（含 $t$） | [傅立葉轉換](Fourier_Transform_to_Shallow_Water_System.ipynb) |
| 第三層 | 正規模態 $\mathbf{K}_{mnr}(\hat{y})$ | $\hat{y}$ | **本篇** |

三層都是「找一組正交完備基底、把方程式對角化」的同一招。第三層之所以做得成，全靠 [能量內積與反厄米性](Energy_Inner_Product_and_Skew_Hermitian_Operator.ipynb) —— 沒有反厄米性，就沒有正交完備的 $\mathbf{K}_{mnr}$，第三層就卡住。

### 受迫阻尼振子的讀法

【證明 (c)】的形狀與最基本的受迫阻尼振子完全一樣：

$$\hat{\eta}_{mnr} = \frac{\overbrace{\kappa\hat{Q}_{mnr}}^{\text{強迫在該模態上的分量}}}{\underbrace{\alpha}_{\text{阻尼}} + i\left(\underbrace{\nu_{mnr}}_{\text{自然頻率}} - \underbrace{\frac{cm}{a}}_{\text{強迫頻率}}\right)}$$

分母的虛部 $\nu_{mnr} - \dfrac{cm}{a}$ 是「失諧量 (detuning)」。它為零時 —— 也就是該波模的**相速** $\dfrac{a\nu_{mnr}}{m}$ 恰好等於對流的**移速** $c$ 時 —— 分母只剩 $\alpha$，響應被放大到【證明 (d)】的最大值。

**這就是「為什麼移動熱源會選擇性地激發特定的赤道波」的全部數學根源。** $c = 5 \ \text{m}\cdot\text{s}^{-1}$ 這個很慢的移速，對相速動輒數十 $\text{m}\cdot\text{s}^{-1}$ 的慣性重力波而言失諧極大（響應被壓抑），對相速慢的羅斯貝波則接近共振。

### 為什麼阻尼一定要在

【假設 2】的 $\alpha > 0$ 不只是數學上的方便。共振寬度由 $\alpha$ 決定：

$$\left|\nu_{mnr} - \frac{cm}{a}\right| \lesssim \alpha \quad \Longrightarrow \quad \text{該模態被有效激發}$$

$\alpha = \left(4 \ \text{days}\right)^{-1} \approx 2.9\times10^{-6} \ \text{s}^{-1}$，而 $\dfrac{c}{a} = \dfrac{5}{6.37\times10^{6}} \approx 7.8\times10^{-7} \ \text{s}^{-1}$。兩者同量級，代表**共振條件相當寬鬆** —— 不必精確對上，一整批相速相近的模態都會被激發。這是 MJO 響應之所以是「一整片波包」而非單一純波的原因。

### 這一篇交付了什麼、還缺什麼

【證明 (c)】給的是**端點①的骨架**：只要知道 $\hat{Q}_{mnr}$，所有譜係數立刻算得出來。

還缺兩塊：

* $\hat{Q}_{mnr}$ 的顯式形式 —— 需要指定加熱的形狀，見 [移動熱源的模態投影](Projection_of_a_Moving_Heat_Source.ipynb)。
* 從譜係數回到物理空間的 $u, v, \phi, q, w$ —— 見 [場還原與 Kelvin 波零 PV](Physical_Field_Recovery_and_Zero_Kelvin_PV.ipynb)。
