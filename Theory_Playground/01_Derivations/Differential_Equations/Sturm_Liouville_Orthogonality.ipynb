# Sturm–Liouville Orthogonality (Sturm–Liouville 本徵函數的正交性與展開係數)

## 證明目標:

* (a) 相異本徵值所對應的本徵函數，在權重 $w(z)$ 之下互相正交：

$$\int_{z_b}^{z_t} \Psi_m(z)\,\Psi_n(z)\,w(z)\,dz = 0 \qquad \left(\lambda_m \neq \lambda_n\right)$$

* (b) 任一可展開的函數 $F(z)$，其第 $m$ 個展開係數（即「投影」）為：

$$q_m = \frac{\displaystyle\int_{z_b}^{z_t} F(z)\,\Psi_m(z)\,w(z)\,dz}{\displaystyle\int_{z_b}^{z_t} \Psi_m^2(z)\,w(z)\,dz}$$

* 註：這是「把一個函數拆解到一組正交基底上」的通用機制。Fourier series 是它在 $p = 1,\ q = 0,\ w = 1$ 與週期邊界下的特例；大氣的垂直模態分解則是它在 $p = 1/N^2,\ q = 0,\ w = 1$ 與剛蓋邊界下的特例，見 [垂直結構方程式 (Vertical Structure Equation)](../Atmospheric_Dynamics/Vertical_Structure_Equation.ipynb)。
* 註：本檔是純數學結果，$p,\ q,\ w,\ \lambda,\ \Psi$ 的因次由套用的物理問題決定，故單位欄一律標記為 $[\text{依應用而定}]$。

## 假設與已知 (Assumptions & Preliminaries)

* **【定義 1】 Sturm–Liouville 算符 (Sturm–Liouville operator)：** 把二階線性常微分算符寫成「自伴形式 (self-adjoint form)」，也就是讓最高階項整個被包進一個全微分裡

  $$\mathcal{L}\left[\Psi\right] \overset{\text{def}}{=} \frac{d}{dz}\left[p(z)\frac{d\Psi}{dz}\right] + q(z)\,\Psi$$

  * $\mathcal{L}$ : Sturm–Liouville 算符 (Sturm–Liouville operator) $[\text{依應用而定}]$
  * $\Psi(z)$ : 本徵函數 (Eigenfunction) $[\text{依應用而定}]$
  * $p(z)$ : 主係數函數 (Leading coefficient function) $[\text{依應用而定}]$
  * $q(z)$ : 零階係數函數 (Zeroth-order coefficient function) $[\text{依應用而定}]$
  * $z$ : 自變數，此處取為垂直座標 (Independent variable) $[\text{m}]$

* **【已知 1】 Sturm–Liouville 本徵值問題 (Sturm–Liouville eigenvalue problem)：** 帶權重的二階線性常微分本徵值問題

  $$\mathcal{L}\left[\Psi_m\right] + \lambda_m\,w(z)\,\Psi_m = 0$$

  把【定義 1】代入即為展開形式

  $$\frac{d}{dz}\left[p(z)\frac{d\Psi_m}{dz}\right] + q(z)\,\Psi_m + \lambda_m\,w(z)\,\Psi_m = 0$$

  * $\Psi_m(z)$ : 第 $m$ 個本徵函數 (The $m$-th eigenfunction) $[\text{依應用而定}]$
  * $\lambda_m$ : 第 $m$ 個本徵值 (The $m$-th eigenvalue) $[\text{依應用而定}]$
  * $w(z)$ : 權重函數 (Weight function) $[\text{依應用而定}]$
  * $m$ : 模態指標 (Mode index) $[\text{無單位}]$

* **【已知 2】 本徵函數的完備性 (Completeness of the eigenfunctions)：** 正則 Sturm–Liouville 問題的本徵函數集 $\left\{\Psi_n\right\}_{n=0}^{\infty}$ 在加權平方可積空間 $L_w^2\left(\left[z_b, z_t\right]\right)$ 中完備，因此任一該空間中的函數 $F(z)$ 皆可展開為

  $$F(z) = \sum_{n=0}^{\infty} q_n\,\Psi_n(z)$$

  * $F(z)$ : 待展開的函數 (Function to be expanded) $[\text{依應用而定}]$
  * $q_n$ : 第 $n$ 個展開係數 (The $n$-th expansion coefficient) $[\text{依應用而定}]$
  * $n$ : 模態指標 (Mode index) $[\text{無單位}]$
  * 註：完備性本身是 Sturm–Liouville 展開定理的結論，屬於泛函分析範疇，本檔只引用不重證。

* **【假設 1】 正則性 (Regularity)：** 本徵值問題定義在有限閉區間 $\left[z_b, z_t\right]$ 上，且係數函數行為良好

  * (a) 係數函數連續：

    $$p,\ \frac{dp}{dz},\ q,\ w \in C\left(\left[z_b, z_t\right]\right)$$

  * (b) 主係數恆正：

    $$p(z) > 0$$

  * (c) 權重恆正：

    $$w(z) > 0$$

  * $z_b$ : 下邊界座標 (Bottom boundary) $[\text{m}]$
  * $z_t$ : 上邊界座標 (Top boundary) $[\text{m}]$
  * 註：(a) 是【證明 (a)】把全微分項還原成邊界值（微積分基本定理）的前提；(b)(c) 則是【已知 2】完備性定理與「本徵值為單重」這件事的前提。

* **【假設 2】 齊次邊界條件 (Homogeneous boundary conditions)：** 上下邊界各給一條齊次條件，三種常見型式擇一即可

  * (a) 第一類（Dirichlet）：

    $$\Psi_m(z_b) = \Psi_m(z_t) = 0$$

  * (b) 第二類（Neumann）：

    $$\left.\frac{d\Psi_m}{dz}\right|_{z_b} = \left.\frac{d\Psi_m}{dz}\right|_{z_t} = 0$$

  * (c) 第三類（Robin），其中 $\beta_b,\ \beta_t \neq 0$：

    $$\alpha_b\,\Psi_m(z_b) + \beta_b\left.\frac{d\Psi_m}{dz}\right|_{z_b} = 0$$

    $$\alpha_t\,\Psi_m(z_t) + \beta_t\left.\frac{d\Psi_m}{dz}\right|_{z_t} = 0$$

  * $\alpha_b,\ \alpha_t$ : 下、上邊界的函數值權重 (Robin coefficients on the value) $[\text{依應用而定}]$
  * $\beta_b,\ \beta_t$ : 下、上邊界的斜率權重 (Robin coefficients on the derivative) $[\text{依應用而定}]$
  * 註：同一個邊界上，**所有**本徵函數共用同一條邊界條件（$\alpha,\ \beta$ 不隨 $m$ 改變），這是【推導 2】能成立的關鍵。

* **【假設 3】 本徵值相異 (Distinct eigenvalues)：** 只比較兩個對應到不同本徵值的本徵函數

  $$\lambda_m \neq \lambda_n$$

  * 註：正則 Sturm–Liouville 問題（【假設 1】）的本徵值必為單重（simple），因此 $m \neq n$ 就足以保證本條成立；若遇到簡併，見【幾何解釋】。

* **【推導 1】 Lagrange 恆等式 (Lagrange's identity)：** 把兩個本徵函數交叉相乘再相減的組合，整理成一個全微分

  $$\begin{gather*}
  \frac{d}{dz}\left[p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right]
  &=& \frac{d}{dz}\left[p\,\Psi_n\frac{d\Psi_m}{dz}\right] - \frac{d}{dz}\left[p\,\Psi_m\frac{d\Psi_n}{dz}\right] \\
  &=& \Psi_n\frac{d}{dz}\left[p\frac{d\Psi_m}{dz}\right] + p\frac{d\Psi_n}{dz}\frac{d\Psi_m}{dz} - \Psi_m\frac{d}{dz}\left[p\frac{d\Psi_n}{dz}\right] - p\frac{d\Psi_m}{dz}\frac{d\Psi_n}{dz} \\
  &=& \Psi_n\frac{d}{dz}\left[p\frac{d\Psi_m}{dz}\right] - \Psi_m\frac{d}{dz}\left[p\frac{d\Psi_n}{dz}\right]
  \end{gather*}$$

* **【推導 2】 邊界項的消失 (Vanishing of the boundary term)：** 【假設 2】的三種型式，任何一種都能讓下面這個 Wronskian 型的組合在該邊界上為零。以下以下邊界 $z_b$ 為例，上邊界 $z_t$ 完全同理

  * (a) 第一類邊界條件

    $$\begin{gather*}
    \left.p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right|_{z_b}
    &\overset{\text{假設 2(a)}}{=}& p(z_b)\left(0 \cdot \left.\frac{d\Psi_m}{dz}\right|_{z_b} - 0 \cdot \left.\frac{d\Psi_n}{dz}\right|_{z_b}\right) \\
    &=& 0
    \end{gather*}$$

  * (b) 第二類邊界條件

    $$\begin{gather*}
    \left.p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right|_{z_b}
    &\overset{\text{假設 2(b)}}{=}& p(z_b)\left(\Psi_n(z_b) \cdot 0 - \Psi_m(z_b) \cdot 0\right) \\
    &=& 0
    \end{gather*}$$

  * (c) 第三類邊界條件

    $$\begin{gather*}
    \left.p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right|_{z_b}
    &\overset{\text{假設 2(c)}}{=}& p(z_b)\left(\Psi_n(z_b)\left(-\frac{\alpha_b}{\beta_b}\Psi_m(z_b)\right) - \Psi_m(z_b)\left(-\frac{\alpha_b}{\beta_b}\Psi_n(z_b)\right)\right) \\
    &=& -p(z_b)\frac{\alpha_b}{\beta_b}\left(\Psi_n(z_b)\Psi_m(z_b) - \Psi_m(z_b)\Psi_n(z_b)\right) \\
    &=& 0
    \end{gather*}$$

  * (d) 因此上下邊界相減後的總邊界項亦為零

    $$\left[p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right]_{z_b}^{z_t} = 0$$

* **【推導 3】 兩條本徵方程式的交叉相減 (Cross-subtraction of the two eigen-equations)：** 用 $\Psi_n$ 乘第 $m$ 條、用 $\Psi_m$ 乘第 $n$ 條，再相減，把 $q(z)$ 項消掉

  $$\begin{gather*}
  0 &\overset{\text{已知 1}}{=}& \Psi_n\left(\mathcal{L}\left[\Psi_m\right] + \lambda_m w\Psi_m\right) - \Psi_m\left(\mathcal{L}\left[\Psi_n\right] + \lambda_n w\Psi_n\right) \\
  &\overset{\text{定義 1}}{=}& \Psi_n\frac{d}{dz}\left[p\frac{d\Psi_m}{dz}\right] + q\,\Psi_n\Psi_m + \lambda_m w\,\Psi_n\Psi_m - \Psi_m\frac{d}{dz}\left[p\frac{d\Psi_n}{dz}\right] - q\,\Psi_m\Psi_n - \lambda_n w\,\Psi_m\Psi_n \\
  &=& \Psi_n\frac{d}{dz}\left[p\frac{d\Psi_m}{dz}\right] - \Psi_m\frac{d}{dz}\left[p\frac{d\Psi_n}{dz}\right] + \left(\lambda_m - \lambda_n\right)w\,\Psi_m\Psi_n \\
  &\overset{\text{推導 1}}{=}& \frac{d}{dz}\left[p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right] + \left(\lambda_m - \lambda_n\right)w\,\Psi_m\Psi_n
  \end{gather*}$$

## 證明:

### (a) proof 本徵函數的正交性 (Orthogonality)

把【推導 3】的結果對整個區間積分，全微分項退化成邊界項，再用【推導 2】把它殺掉。

$$\begin{gather*}
0 &\overset{\text{推導 3}}{=}& \int_{z_b}^{z_t}\left(\frac{d}{dz}\left[p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right] + \left(\lambda_m - \lambda_n\right)w\,\Psi_m\Psi_n\right)dz \\
0 &\overset{\text{假設 1(a)}}{=}& \left[p\left(\Psi_n\frac{d\Psi_m}{dz} - \Psi_m\frac{d\Psi_n}{dz}\right)\right]_{z_b}^{z_t} + \left(\lambda_m - \lambda_n\right)\int_{z_b}^{z_t} w\,\Psi_m\Psi_n\,dz \\
0 &\overset{\text{推導 2(d)}}{=}& \left(\lambda_m - \lambda_n\right)\int_{z_b}^{z_t} w\,\Psi_m\Psi_n\,dz \\
0 &\overset{\text{假設 3}}{=}& \int_{z_b}^{z_t} \Psi_m(z)\,\Psi_n(z)\,w(z)\,dz
\end{gather*}$$

### (b) proof 展開係數（投影公式）(Expansion coefficient)

$$\begin{gather*}
\int_{z_b}^{z_t} F\,\Psi_m\,w\,dz &\overset{\text{已知 2}}{=}& \int_{z_b}^{z_t}\left(\sum_{n=0}^{\infty} q_n\Psi_n\right)\Psi_m\,w\,dz \\
\int_{z_b}^{z_t} F\,\Psi_m\,w\,dz &=& \sum_{n=0}^{\infty} q_n\int_{z_b}^{z_t} \Psi_n\Psi_m\,w\,dz \\
\int_{z_b}^{z_t} F\,\Psi_m\,w\,dz &\overset{\text{證明 (a)}}{=}& q_m\int_{z_b}^{z_t} \Psi_m^2\,w\,dz \\
q_m &=& \frac{\displaystyle\int_{z_b}^{z_t} F(z)\,\Psi_m(z)\,w(z)\,dz}{\displaystyle\int_{z_b}^{z_t} \Psi_m^2(z)\,w(z)\,dz}
\end{gather*}$$

## 幾何解釋

### 一切都只是「向量在正交基底上取分量」

定義加權內積

$$\left\langle f,\ g\right\rangle_w \overset{\text{def}}{=} \int_{z_b}^{z_t} f(z)\,g(z)\,w(z)\,dz$$

那麼 (a) 就是 $\left\langle \Psi_m,\ \Psi_n\right\rangle_w = 0$，(b) 就是

$$q_m = \frac{\left\langle F,\ \Psi_m\right\rangle_w}{\left\langle \Psi_m,\ \Psi_m\right\rangle_w}$$

這跟三維空間裡把向量 $\vec{F}$ 拆成 $\hat{e}_x,\ \hat{e}_y,\ \hat{e}_z$ 的分量 $F_x = \vec{F}\cdot\hat{e}_x$ 是**同一件事**，只是把「三個分量的求和」換成「對 $z$ 的積分」。若把本徵函數正規化成 $\left\langle \Psi_m, \Psi_m\right\rangle_w = 1$，分母就消失，投影公式退化成最乾淨的 $q_m = \left\langle F, \Psi_m\right\rangle_w$。

### $w(z)$ 是內積的「度規」，不是可有可無的裝飾

正交性**只在 $w(z)$ 這個權重下成立**。換一個權重去積分，$\int \Psi_m\Psi_n\,dz$ 一般不會是零，投影就會「洩漏」到別的模態上。所以每次做模態分解，第一件事是先問清楚：**這個問題的 $p$、$w$ 分別是誰？** 在大氣的垂直模態問題裡，$p = 1/N^2$、$w = 1$，因此正交關係恰好是最單純的 $\int \Psi_m\Psi_n\,dz = 0$。

### 為什麼需要「本徵值相異」

(a) 的最後一步是把 $\left(\lambda_m - \lambda_n\right)$ 除掉，這一步在 $\lambda_m = \lambda_n$（簡併，degeneracy）時是非法的。簡併時本徵函數所張的是一個多維子空間，子空間**內部**的基底並不自動正交，必須另外用 Gram–Schmidt 正交化。所幸正則 Sturm–Liouville 問題（【假設 1】）的本徵值必定是單重的（simple），所以這裡不會遇到這個麻煩。

### 邊界條件不是配角，是主角

(a) 真正的力學來源是【推導 2】：**邊界項必須為零**。這就是為什麼「同一組邊界條件」是模態分解能成立的前提 —— 換了邊界條件，就是換了一整套 $\Psi_m$ 與 $\lambda_m$，先前算出來的投影全部作廢。這一點在物理問題上特別容易被忽略：$\Psi_0$ 到底長什麼樣、$\lambda_0$ 是多少，答案完全被上下邊界綁死。
